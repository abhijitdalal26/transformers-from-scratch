# minbpe — Hands-on Experiments

Karpathy's BPE tokenizer library. Run the cells in order.

**Note:** Run this notebook from the `minbpe` folder (so `import minbpe` works). Select the `venv` kernel, or install the package first with `pip install -e .`.

## 1. Train a tokenizer and try it out

Train a `RegexTokenizer` (GPT-4-style pre-tokenization) on the Taylor Swift text, then encode/decode some text and inspect what the tokenizer learned.

```python
from minbpe import BasicTokenizer, RegexTokenizer, GPT4Tokenizer

text = open("tests/taylorswift.txt", encoding="utf-8").read()
print(f"Training text: {len(text):,} characters")

tok = RegexTokenizer()
tok.train(text, vocab_size=512, verbose=True)
```

```python
# Try it out
ids = tok.encode("Hello, Taylor Swift!")
print("token ids:", ids)
print("round-trip:", tok.decode(ids))
```

```python
# See what it learned (merged tokens, sorted by id)
# tok.vocab is a dict: token id -> bytes; ids >= 256 are learned merges
print(f"vocab size: {len(tok.vocab)}\n")
print("last 30 learned tokens:")
for i in sorted(tok.vocab)[-30:]:
    print(f"  {i:4d}: {tok.vocab[i]!r}")
```

## 2. Use the saved models (no retraining)

The models trained by `train.py` are saved in `models/` — load them directly.

```python
tok2 = RegexTokenizer()
tok2.load("models/regex.model")

ids = tok2.encode("Hello, Taylor Swift! 2024")
print("token ids:", ids)
print("round-trip:", tok2.decode(ids))
```

## 3. Vocabulary size trade-off

Bigger vocab = fewer tokens per text, but a larger tokenizer/model. See it in action. (vocab size must be ≥ 256 since BPE starts from all 256 bytes)

```python
msg = "Taylor Swift released a new album in 2024"

print(f"{'vocab size':>10} | {'tokens':>6} | ids")
print("-" * 80)
# vocab sizes must be >= 256 (base vocabulary is all 256 bytes)
for size in (256, 384, 512, 1024):
    t = RegexTokenizer()
    t.train(text, size, verbose=False)
    ids = t.encode(msg)
    print(f"{size:>10} | {len(ids):>6} | {ids}")
```

## 4. Bonus: compare the three tokenizers

- **BasicTokenizer** — plain byte-level BPE, no pre-tokenization
- **RegexTokenizer** — GPT-4-style regex pre-tokenization
- **GPT4Tokenizer** — matches real GPT-4 (tiktoken) exactly

```python
msg = "Hello, Taylor Swift! 안녕하세요 😊"

# Basic and Regex tokenizers are trained on our text;
# GPT4Tokenizer is pretrained (loads GPT-4's real vocab from tiktoken) - no training needed
for cls in (BasicTokenizer, RegexTokenizer, GPT4Tokenizer):
    t = cls()
    if cls is not GPT4Tokenizer:
        t.train(text, 512, verbose=False)
    ids = t.encode(msg)
    print(f"{cls.__name__:18} ({len(ids):2d} tokens): {ids}")

# sanity check: GPT4Tokenizer round-trips correctly
gpt4 = GPT4Tokenizer()
assert gpt4.decode(gpt4.encode(msg)) == msg
print("\nGPT4Tokenizer round-trip OK")
```
